```python
from transformers import CLIPModel
import torch

# model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")

model = CLIPModel.from_pretrained("NemesisAlm/clip-fine-tuned-satellite")
```

```python
batch_size = 1
num_channels = 3
height = 224
width = 224

dummy_input = torch.randn(batch_size, num_channels, height, width)
```

```python
import torch.onnx

torch.onnx.export(model.vision_model,
                  dummy_input,
                  "clip_satellite_visual_encoder.onnx",
                  export_params=True,
                  opset_version=17,
                  do_constant_folding=True,
                  input_names=['input'],
                  output_names=['output'],
                  dynamic_axes={'input': {0: 'batch_size'},
                                'output': {0: 'batch_size'}})
```
